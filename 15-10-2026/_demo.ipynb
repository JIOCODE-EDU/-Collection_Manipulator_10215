#### Python List vs Numpy Array

```python
python_list = [1 , 2 , 3 , 4 , 5]

doubled_list = [x * 2 for x in python_list]

print(doubled_list)

print("\n=====  Numpy Array ======\n")

numpy_array = np.array([1 , 2 , 3 , 4 , 5])

doubled_array = numpy_array * 2

print(doubled_array)

```
